```python
import pandas as pd
```

```python
df_sales1 = pd.read_csv("../data/Sales_MedicStore_part1.csv", sep=";")
```

```python
df_sales2 = pd.read_csv("../data/Sales_MedicStore_part2.csv", sep=";")
```

```python
df_mapping = pd.read_excel("../data/Mapping.xlsx")
```

```python
df_sales1.head()
```

```python
df_sales2.head()
```

```python
df_mapping.head()
```

```python
lst_sales = [df_sales1, df_sales2]
df_sales = pd.concat(lst_sales)
```

```python
print("shape of df_sales1:", df_sales1.shape)
print("shape of df_sales2:", df_sales2.shape)
print("shape of df_sales:", df_sales.shape)
```

```python
df_sales_mapping = pd.merge(df_sales, df_mapping, left_on="Sub-Category", right_on="Sub-Category")
```

```python
df_sales_mapping.shape
```

```python
df_sales_mapping.head()
```

```python
df_sales_mapping["total_sales"] = df_sales_mapping["Quantity"] * df_sales_mapping["Unit Price"]
```

```python
df_sales_mapping[["Quantity", "Unit Price", "total_sales"]].head()
```

```python
df_sales_mapping.to_excel("../data/sales_consolidated.xlsx", index=False)
```

```python
# Extract the sales made in France
df_sales_france = df_sales_mapping[df_sales_mapping["Country"] == "France"]

# Business Q4: total sales for France
print(df_sales_france["total_sales"].sum())
```

```python
# Total sales for France
sales_france = df_sales_france["total_sales"].sum()

# Total sales for all countries
sales_total = df_sales_mapping["total_sales"].sum()

# Business Q5: share of France in the total sales, in percent
share_france = sales_france / sales_total * 100
print(share_france)
```
